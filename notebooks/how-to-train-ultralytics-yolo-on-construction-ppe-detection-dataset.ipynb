# Construction PPE Detection using Ultralytics YOLO26

This notebook serves as an initial step for training the YOLO26 model on the [construction-ppe](https://docs.ultralytics.com/datasets/detect/construction-ppe/) detection dataset.

## Dataset Structure

The construction-ppe dataset is divided into three subsets:

- **Training set**: Consisting of 1132 images, each accompanied by corresponding annotations.
- **Validation set**: Comprising 143 images, with annotations paired for each one.
- **Testing set**: Comprising 141 images, with annotations paired for each one.

## Applications

Construction-PPE detection using computer vision improves workplace safety and compliance. By analyzing site footage, AI systems check if workers wear helmets, vests, or gloves, enabling real-time monitoring and instant alerts. These systems can also provide safety analytics, support drone or robot inspections, and serve as valuable tools for research and education in workplace safety.

````
import os
import shutil
from pathlib import Path
import yaml

def convert_sh17_to_yolo(dataset_path, output_path="datasets/sh17_yolo"):
    """
    Convert SH17 PPE dataset to YOLO format

    Expected SH17 structure (common formats):
    - images/ (or train/, val/, test/)
    - annotations/ (XML, JSON, or txt)

    YOLO structure:
    - images/train/
    - images/val/
    - labels/train/
    - labels/val/
    - data.yaml
    """

    dataset_path = Path(dataset_path)
    output_path = Path(output_path)

    # Create YOLO directory structure
    for split in ['train', 'val']:
        (output_path / 'images' / split).mkdir(parents=True, exist_ok=True)
        (output_path / 'labels' / split).mkdir(parents=True, exist_ok=True)

    # Define PPE classes (adjust based on actual dataset)
    # Common SH17 classes: helmet, vest, person, no-helmet, no-vest
    classes = {
        'helmet': 0,
        'vest': 1,
        'no-helmet': 2,
        'no-vest': 3,
        'person': 4
    }

    print(f"Converting dataset from {dataset_path}")

    # Check dataset format
    if (dataset_path / 'annotations').exists():
        annotations_path = dataset_path / 'annotations'
        images_path = dataset_path / 'images'
    else:
        # Might be in train/val structure already
        annotations_path = dataset_path
        images_path = dataset_path

    # Process based on annotation format
    annotation_files = list(annotations_path.rglob("*.txt")) or \
                      list(annotations_path.rglob("*.xml")) or \
                      list(annotations_path.rglob("*.json"))

    print(f"Found {len(annotation_files)} annotation files")

    # If already in YOLO format, just copy
    if annotation_files and annotation_files[0].suffix == '.txt':
        # Check if it's already YOLO format
        with open(annotation_files[0]) as f:
            first_line = f.readline().strip()
            if first_line and len(first_line.split()) == 5:
                print("Dataset appears to be in YOLO format already")
                # Copy files maintaining train/val split if exists
                copy_yolo_dataset(dataset_path, output_path)
                create_data_yaml(output_path, classes)
                return output_path

    # Create data.yaml
    create_data_yaml(output_path, classes)

    return output_path


def copy_yolo_dataset(src_path, dst_path):
    """Copy already formatted YOLO dataset"""
    src_path = Path(src_path)
    dst_path = Path(dst_path)

    # Look for train/val structure
    for split in ['train', 'val', 'test']:
        src_images = src_path / 'images' / split
        src_labels = src_path / 'labels' / split

        if src_images.exists():
            dst_images = dst_path / 'images' / split
            dst_labels = dst_path / 'labels' / split

            dst_images.mkdir(parents=True, exist_ok=True)
            dst_labels.mkdir(parents=True, exist_ok=True)

            # Copy images
            for img in src_images.glob("*.[jp][pn][g]"):
                shutil.copy2(img, dst_images / img.name)

            # Copy labels
            if src_labels.exists():
                for label in src_labels.glob("*.txt"):
                    shutil.copy2(label, dst_labels / label.name)

            print(f"Copied {split} split: {len(list(dst_images.glob('*')))} images")


def create_data_yaml(output_path, classes):
    """Create YOLO data.yaml configuration"""
    data_config = {
        'path': str(output_path.absolute()),
        'train': 'images/train',
        'val': 'images/val',
        'nc': len(classes),
        'names': list(classes.keys())
    }

    yaml_path = output_path / 'data.yaml'
    with open(yaml_path, 'w') as f:
        yaml.dump(data_config, f, default_flow_style=False)

    print(f"Created {yaml_path}")
    return yaml_path


# Convert the dataset
yolo_dataset_path = convert_sh17_to_yolo(path)
print(f"\nYOLO dataset ready at: {yolo_dataset_path}")
````

````
import shutil
from pathlib import Path
import yaml

def combine_ppe_datasets_with_remapping(
    construction_ppe_path="datasets/construction-ppe",
    sh17_path="datasets/sh17_yolo",
    output_path="datasets/ppe_combined"
):
    """
    Combine construction-ppe and sh17_yolo with unified class mapping
    """

    construction_ppe_path = Path(construction_ppe_path)
    sh17_path = Path(sh17_path)
    output_path = Path(output_path)

    # === UNIFIED CLASS MAPPING ===
    # Create a unified class list
    unified_classes = [
        'helmet',           # 0
        'no-helmet',        # 1
        'vest',             # 2
        'no-vest',          # 3
        'person',           # 4
        'gloves',           # 5
        'no-gloves',        # 6
        'boots',            # 7
        'no-boots',         # 8
        'goggles',          # 9
        'no-goggles',       # 10
    ]

    # SH17 mapping (already matches our unified classes 0-4)
    sh17_to_unified = {
        0: 0,  # helmet -> helmet
        1: 2,  # vest -> vest
        2: 1,  # no-helmet -> no-helmet
        3: 3,  # no-vest -> no-vest
        4: 4,  # person -> person
    }

    # Construction-PPE mapping
    construction_to_unified = {
        0: 0,   # helmet -> helmet
        1: 5,   # gloves -> gloves
        2: 2,   # vest -> vest
        3: 7,   # boots -> boots
        4: 9,   # goggles -> goggles
        5: -1,  # none -> skip (special handling)
        6: 4,   # Person -> person
        7: 1,   # no_helmet -> no-helmet
        8: 10,  # no_goggle -> no-goggles
        9: 6,   # no_gloves -> no-gloves
        10: 8,  # no_boots -> no-boots
    }

    # Create output directory structure
    for split in ['train', 'val', 'test']:
        (output_path / 'images' / split).mkdir(parents=True, exist_ok=True)
        (output_path / 'labels' / split).mkdir(parents=True, exist_ok=True)

    print("="*60)
    print("COMBINING PPE DATASETS WITH CLASS REMAPPING")
    print("="*60)

    # === Process SH17 dataset ===
    print("\n📦 Processing SH17 dataset...")
    sh17_stats = process_dataset_with_remap(
        src_path=sh17_path,
        dst_path=output_path,
        class_mapping=sh17_to_unified,
        prefix="sh17",
        splits=['train', 'val']
    )

    # === Process construction-ppe dataset ===
    print("\n📦 Processing construction-ppe dataset...")
    const_stats = process_dataset_with_remap(
        src_path=construction_ppe_path,
        dst_path=output_path,
        class_mapping=construction_to_unified,
        prefix="const",
        splits=['train', 'val', 'test']
    )

    # === Create unified data.yaml ===
    create_unified_data_yaml(output_path, unified_classes)

    # === Print Statistics ===
    print("\n" + "="*60)
    print("COMBINED DATASET STATISTICS")
    print("="*60)

    for split in ['train', 'val', 'test']:
        img_count = len(list((output_path / 'images' / split).glob("*")))
        lbl_count = len(list((output_path / 'labels' / split).glob("*.txt")))
        print(f"{split.upper():6s}: {img_count:5d} images, {lbl_count:5d} labels")

    print("\n" + "="*60)
    print("CLASS DISTRIBUTION")
    print("="*60)
    analyze_class_distribution(output_path, unified_classes)

    return output_path


def process_dataset_with_remap(src_path, dst_path, class_mapping, prefix, splits):
    """
    Process a dataset with class remapping
    """
    stats = {split: {'images': 0, 'labels': 0, 'boxes': 0} for split in splits}

    for split in splits:
        src_images = src_path / 'images' / split
        src_labels = src_path / 'labels' / split

        if not src_images.exists():
            print(f"  ⚠️  {split} split not found, skipping")
            continue

        dst_images = dst_path / 'images' / split
        dst_labels = dst_path / 'labels' / split

        for img in src_images.glob("*"):
            if img.suffix.lower() not in ['.jpg', '.jpeg', '.png', '.bmp']:
                continue

            # Copy image with prefix
            dst_img = dst_images / f"{prefix}_{img.name}"
            shutil.copy2(img, dst_img)
            stats[split]['images'] += 1

            # Process and remap label
            src_label = src_labels / f"{img.stem}.txt"
            if src_label.exists():
                dst_label = dst_labels / f"{prefix}_{img.stem}.txt"
                boxes_written = remap_and_copy_label(
                    src_label,
                    dst_label,
                    class_mapping
                )
                if boxes_written > 0:
                    stats[split]['labels'] += 1
                    stats[split]['boxes'] += boxes_written

        print(f"  ✓ {split:5s}: {stats[split]['images']:4d} images, "
              f"{stats[split]['labels']:4d} labels, "
              f"{stats[split]['boxes']:5d} boxes")

    return stats


def remap_and_copy_label(src_file, dst_file, class_mapping):
    """
    Remap class IDs and copy label file
    Returns number of boxes written
    """
    remapped_lines = []

    with open(src_file) as f:
        for line in f:
            parts = line.strip().split()
            if not parts:
                continue

            old_class = int(parts[0])

            # Skip if class should be ignored (-1 mapping)
            if old_class not in class_mapping or class_mapping[old_class] == -1:
                continue

            # Remap class ID
            new_class = class_mapping[old_class]
            parts[0] = str(new_class)
            remapped_lines.append(' '.join(parts))

    # Write remapped labels
    if remapped_lines:
        with open(dst_file, 'w') as f:
            f.write('\n'.join(remapped_lines) + '\n')

    return len(remapped_lines)


def create_unified_data_yaml(output_path, classes):
    """Create unified data.yaml"""

    data_config = {
        'path': str(output_path.absolute()),
        'train': 'images/train',
        'val': 'images/val',
        'test': 'images/test',
        'nc': len(classes),
        'names': classes
    }

    yaml_path = output_path / 'data.yaml'
    with open(yaml_path, 'w') as f:
        yaml.dump(data_config, f, default_flow_style=False, sort_keys=False)

    print(f"\n✓ Created {yaml_path}")
    print(f"  Classes: {classes}")

    # Create classes reference file
    with open(output_path / 'classes.txt', 'w') as f:
        for i, name in enumerate(classes):
            f.write(f"{i}: {name}\n")


def analyze_class_distribution(dataset_path, class_names):
    """Analyze class distribution across the combined dataset"""
    from collections import defaultdict

    class_counts = defaultdict(int)
    total_boxes = 0

    dataset_path = Path(dataset_path)

    for split in ['train', 'val', 'test']:
        label_dir = dataset_path / 'labels' / split
        if not label_dir.exists():
            continue

        for label_file in label_dir.glob("*.txt"):
            with open(label_file) as f:
                for line in f:
                    parts = line.strip().split()
                    if parts:
                        class_id = int(parts[0])
                        class_counts[class_id] += 1
                        total_boxes += 1

    # Print distribution
    for class_id in sorted(class_counts.keys()):
        count = class_counts[class_id]
        percentage = (count / total_boxes * 100) if total_boxes > 0 else 0
        class_name = class_names[class_id] if class_id < len(class_names) else f"unknown_{class_id}"
        print(f"  {class_id:2d}. {class_name:15s}: {count:6d} boxes ({percentage:5.2f}%)")

    print(f"\n  Total boxes: {total_boxes}")


# === RUN THE COMBINATION ===
combined_path = combine_ppe_datasets_with_remapping(
    construction_ppe_path="datasets/construction-ppe",
    sh17_path="datasets/sh17_yolo",
    output_path="datasets/ppe_combined"
)

print(f"\n{'='*60}")
print("✓ DATASET COMBINATION COMPLETE!")
print(f"{'='*60}")
print(f"📁 Location: {combined_path}")
print(f"📄 Config: {combined_path}/data.yaml")
print(f"\n🚀 Ready to train:")
print(f"   yolo train data={combined_path}/data.yaml model=yolov11n.pt epochs=100")
````

````
# ========================================
# STEP 1: RESTART COLAB RUNTIME FIRST!
# ========================================
# Go to: Runtime -> Restart runtime
# Then run the cells below

# ========================================
# STEP 2: Install and Setup
# ========================================
!pip install ultralytics -q

import torch
import gc
import os

# Set memory configuration BEFORE any CUDA operations
os.environ['PYTORCH_CUDA_ALLOC_CONF'] = 'expandable_segments:True'

# Clear any existing cache
torch.cuda.empty_cache()
gc.collect()

# Verify GPU
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"Total GPU memory: {torch.cuda.get_device_properties(0).total_memory/1024**3:.2f} GB")

# ========================================
# STEP 3: Train with Memory-Safe Settings
# ========================================
from ultralytics import YOLO

model = YOLO('yolo26m.pt')  # Your YOLO26m model

results = model.train(
    # Data
    data='datasets/ppe_combined/data.yaml',

    # Core settings - OPTIMIZED FOR T4
    epochs=200,
    imgsz=640,
    batch=8,  # ⭐ REDUCED to 8 for YOLO26m (21M params)
    name='ppe_yolo26m_t4_safe',
    device=0,

    # Memory optimization
    workers=2,  # ⭐ Reduced workers
    cache=False,  # ⭐ DISABLE cache to save VRAM
    amp=True,

    # Training
    patience=50,
    save=True,
    save_period=20,  # Save less frequently
    plots=True,

    # Reduced augmentation to save memory
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=10.0,  # Reduced from 15
    translate=0.1,
    scale=0.5,
    shear=1.0,  # Reduced from 2.0
    perspective=0.0,
    flipud=0.0,
    fliplr=0.5,
    mosaic=1.0,
    mixup=0.0,  # ⭐ DISABLED
    copy_paste=0.0,  # ⭐ DISABLED

    # Optimizer - SGD uses less memory than AdamW
    optimizer='SGD',  # ⭐ Changed from AdamW
    lr0=0.01,
    lrf=0.01,
    momentum=0.937,
    weight_decay=0.0005,
    warmup_epochs=3.0,
    warmup_momentum=0.8,

    # Other
    close_mosaic=10,
)

print("\n✓ Training complete!")
print(f"📁 Results: {results.save_dir}")
````
